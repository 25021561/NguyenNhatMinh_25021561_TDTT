```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# 1. Khai báo thông số đề bài
n = 100      # Số lượng linh kiện kiểm tra
p = 0.05     # Tỷ lệ lỗi (5%)

# 2. Câu 1: Xác suất có ĐÚNG 3 linh kiện lỗi P(X = 3)
# Hàm pmf (Probability Mass Function) tính P(X = k)
prob_exact_3 = binom.pmf(3, n, p)

# 3. Câu 2: Xác suất có TỐI ĐA 5 linh kiện lỗi P(X <= 5)
# Hàm cdf (Cumulative Distribution Function) tính P(X <= k)
prob_max_5 = binom.cdf(5, n, p)

# In kết quả ra màn hình
print("=" * 45)
print(f"1. Xác suất có ĐÚNG 3 linh kiện lỗi : {prob_exact_3:.5f} ({prob_exact_3*100:.2f}%)")
print(f"2. Xác suất có TỐI ĐA 5 linh kiện lỗi: {prob_max_5:.5f} ({prob_max_5*100:.2f}%)")
print("=" * 45)

# 4. Câu 3: Vẽ đồ thị trực quan hóa phân phối xác suất
x = np.arange(0, 16)  # Khảo sát từ 0 đến 15 sản phẩm lỗi
y = binom.pmf(x, n, p)

plt.figure(figsize=(9, 5))
plt.bar(x, y, color='skyblue', edgecolor='navy', alpha=0.7, label='P(X = k)')
plt.bar(3, binom.pmf(3, n, p), color='crimson', label='Đúng 3 lỗi (Câu 1)')

plt.title('Phân Phối Xác Suất Số Linh Kiện Lỗi Trong Lô 100 Sản Phẩm', fontsize=12)
plt.xlabel('Số linh kiện bị lỗi (k)')
plt.ylabel('Xác suất P(X = k)')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-8, 8, 100)
y = np.linspace(-8, 8, 100)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2) + 1e-10
Z = np.sin(R) / R
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='jet')
fig.colorbar(surf)
ax.set_title("Đồ thị 3D trong Python: z = sin(r)/r")
plt.show()
```
